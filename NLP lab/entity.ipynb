```python
import spacy

nlp = spacy.load("en_core_web_sm")

text = """My name is Narayan Das I belong to Mumbai
Computer science and ML subject in Amity University
"""

doc = nlp(text)

print("-------Named Entity Recognition (NER) Output-------\n")

for ent in doc.ents:
    print("Entity Text:", ent.text)
    print("Entity Label:", ent.label_)
    print("------------------")


text2 = "My name is Liju Dass"

doc2 = nlp(text2)

print("\n-------Named Entity Recognition (NER) Output-------\n")

for ent in doc2.ents:
    print("Entity Text:", ent.text)
    print("Entity Label:", ent.label_)
    print("------------------")
```
